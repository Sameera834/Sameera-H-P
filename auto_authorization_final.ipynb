```python
import face_recognition
import cv2
import time

# Load the input image
input_image = face_recognition.load_image_file('sam.jpg')
input_encoding = face_recognition.face_encodings(input_image)[0]

# Initialize the camera
cap = cv2.VideoCapture(0)

start_time = time.time()
last_detection_time = time.time()
no_match_time = time.time()

while True:
    # Read a frame from the camera
    ret, frame = cap.read()

    # Convert the frame to RGB
    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)

    # Detect faces in the frame
    face_locations = face_recognition.face_locations(rgb_frame)

    # Compute the face encodings
    face_encodings = face_recognition.face_encodings(rgb_frame, face_locations)

    # Check if 10 seconds have not passed since the program started
    if time.time() - start_time <= 10:
        # Loop through each face location and encoding
        for face_location, face_encoding in zip(face_locations, face_encodings):
            # Compare the face encoding with the input encoding
            matches = face_recognition.compare_faces([input_encoding], face_encoding)

            # Check if the face is recognized
            if matches[0]:
                print("Person detected")
                top, right, bottom, left = face_location
                cv2.rectangle(frame, (left, top), (right, bottom), (0, 255, 0), 2)
                # Create a new window to display the detected face
                cv2.namedWindow('Detected Face', cv2.WINDOW_NORMAL)
                cv2.imshow('Detected Face', frame)
                last_detection_time = time.time()
                no_match_time = time.time()
                break
        else:
            print("No match found")
            if time.time() - no_match_time > 10:
                print("No match found for 10 seconds. Exiting program.")
                break
    # Check if 10 seconds have passed since the program started
    elif time.time() - start_time > 10:
        # Check if 900 seconds have passed since the last detection
        if time.time() - last_detection_time > 20:
            # Loop through each face location and encoding
            for face_location, face_encoding in zip(face_locations, face_encodings):
                # Compare the face encoding with the input encoding
                matches = face_recognition.compare_faces([input_encoding], face_encoding)

                # Check if the face is recognized
                if matches[0]:
                    print("Person detected")
                    top, right, bottom, left = face_location
                    cv2.rectangle(frame, (left, top), (right, bottom), (0, 255, 0), 2)
                    # Create a new window to display the detected face
                    cv2.namedWindow('Detected Face', cv2.WINDOW_NORMAL)
                    cv2.imshow('Detected Face', frame)
                    last_detection_time = time.time()
                    no_match_time = time.time()
                    break
            else:
                print("No match found")
                if time.time() - no_match_time > 10:
                    print("No match found for 10 seconds. Exiting program.")
                    break
        else:
            print("Checking again in {:.2f} minutes".format((20 - (time.time() - last_detection_time)) / 60))

    # Display the frame
    cv2.imshow('Face Recognition', frame)

    # Check for the 'q' key to exit the program
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release the camera and close all windows
cap.release()
cv2.destroyAllWindows()
```
